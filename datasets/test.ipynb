```python
import pandas as pd
```

```python
PATH_DATASET = 'page_blocks'
```

```python
import subprocess
import io

# .Z is Unix LZW compress format — not supported by pandas directly.
# uncompress -c decompresses to stdout without touching the original file.
result = subprocess.run(
    ["uncompress", "-c", f"./{PATH_DATASET}/page-blocks.data.Z"],
    capture_output=True,
    check=True,
)

# Per page-blocks.names: 10 numeric attributes + 1 class label.
# Classes (1-based in the source file):
#   1 = text, 2 = horiz. line, 3 = picture, 4 = vert. line, 5 = graphic
columns = [
    "height",
    "lenght",
    "area",
    "eccen",
    "p_black",
    "p_and",
    "mean_tr",
    "blackpix",
    "blackand",
    "wb_trans",
    "class",
]

df = pd.read_csv(
    io.BytesIO(result.stdout),
    sep=r"\s+",
    header=None,
    names=columns,
)
df
```

```python
df.shape, list(df.columns)
```

```python
df['class'].value_counts()
```

```python
unique_labels = sorted(df["class"].unique())
label_to_index = {label: i for i, label in enumerate(unique_labels)}
df["class"] = df["class"].map(label_to_index).astype(int)

print(f"Classes: {len(unique_labels)} (mapping {label_to_index})")
df["class"].value_counts().sort_index()
```

```python
df.to_csv(f"{PATH_DATASET}.csv", index=False)
```
