```python
import pandas as pd
```

```python
PATH_DATASET = 'waveform'
```

```python
import subprocess
import io

# .Z is Unix LZW compress format — not supported by pandas directly.
# uncompress -c decompresses to stdout without touching the original file.
result = subprocess.run(
    ["uncompress", "-c", f"./{PATH_DATASET}/waveform.data.Z"],
    capture_output=True,
    check=True,
)

# 40 continuous attributes + class (0 / 1 / 2)
columns = [f"attr{i}" for i in range(1, 21)] + ["class"]

df = pd.read_csv(
    io.BytesIO(result.stdout),
    sep=",",
    header=None,
    names=columns,
)
df
```

```python
len(df.columns)
```

```python
df['class'].value_counts()
```

```python
df = df[df["class"].isin([0, 1])].reset_index(drop=True)
df
```

```python
df.to_csv(f"{PATH_DATASET}.csv", index=False)
```
